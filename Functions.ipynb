Playing with parameters

```python
# Addition of two numbers

def add_two_num(val1,val2):
  a1 = val1
  print("value of a1 -",a1)
  a2 = val2
  print("value of a2 -",a2)
  a3 = a1 + a2
  print("Addition of a1 and a2 -",a3)
```

```python
add_two_num(1,2)  # Positional arguments
```

```python
add_two_num(20,10)
```

```python
add_two_num(val2=20,val1=10)   # Keyword arguments
```

```python
# Addition of two numbers with default parameter

def add_two_num(val1,val2=10):    #default paramaters
  a1 = val1
  print("value of a1 -",a1)
  a2 = val2
  print("value of a2 -",a2)
  a3 = a1 + a2
  print("Addition of a1 and a2 -",a3)
```

```python
add_two_num(20)
```

```python
add_two_num(20,40)  # value 40 overrides default parameter value
```

```python
def add_two_num(val1=10,val2):    #default paramaters
  a1 = val1
  print("value of a1 -",a1)
  a2 = val2
  print("value of a2 -",a2)
  a3 = a1 + a2
  print("Addition of a1 and a2 -",a3)
```

**Return Keyword**

```python
# Multiplication of two numbers

def mul_two_num(v1,v2):
  print("user given value 1 -",v1)
  print("user given value 2 -",v2)
  m1 = v1 * v2
  print("Multiplication of v1 and v2 -",m1)
  return m1
```

```python
m = mul_two_num(10,20)
```

write a function that returns the total cost of the purchase. Parameters required are price and quantity

```python
def total_cost(n1,n2):
  price = n1
  print("price of the product -",n1)
  quantity = n2
  print("quantity of the product -",n2)
  cost = price * quantity
  print("Total cost is ",cost)
  return cost
```

```python
c = total_cost(100,2)
```
