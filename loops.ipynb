# ***While loop***

```python
#initialization, creation, increment/decrement

# Print values 1 to 20
c=1
while c <= 20:
  print(c)
  c+=1
```

```python
# Print value from 20 to 1

d=20
while d>=1:
  print(d)
  d-=1
```

# *For Loop*

```python
# For loop with list

fruits = ['apple','banana','cherry']
```

```python
fruits[1]
```

```python
for i in fruits:
  print(i)
```

**For loop with range**

```python
# Print values from 1 to 10

for i in range(1, 11, 1):
  print(i)
```

```python
# print all even numbers between 1 to 20

for j in range(2, 21, 2):
  print(j)
```

```python
# print from 1 to 10

for z in range(11):
  print(z)
```

```python
# Range default starting value will be 0
# range default increment value will be 1

for q in range(6):
  print(q)
```

```python
# print 22 to 11 in reverse order

for w in range (22, 10, -1):
  print(w)
```

***For loop with conditional statements***

```python
for i in range(1,11,1):
  print(i)
```

```python
for i in range(1,11):
  #print(i)
  if (i % 2 == 0):
    print("Even")
  else:
    print("Odd")
```

```python
# print from 1 to 10 but i don't need 5

for i in range(1,11):
  if (i != 5):
    print(i)
```

```python
# print from 1 to 10 without 5 ( Using pass function)

for a in range(1,6):
  if (a == 5):
    pass
  else:
    print(a)
```

```python
# print 1 to 6 without 5 ( using continue( skip )) Continue won't stop loop condition if condition satisfies and it will continue to run till the stop range

for i in range(1,7):
  if (i == 5):
    continue
  else:
    print(i)
```

```python
# print 5,15,20,25,30

for i in range(5,35,5):
  if (i == 10):
    continue
  else:
   print(i)
```

```python
# print 5,15,20,25,30 ( using break function ) break will stop the loop condition if condition satisfies
for i in range(5,35,5):
  if (i == 10):
    break
  else:
   print(i)
```

```python
# Run loop from 1 to 100 but it need to stop at 20

for i in range(1,101):
  if (i > 20):
    break
  print(i)
```
