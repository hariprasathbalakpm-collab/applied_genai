# **If Else Statement**

```python
if age >=18:
  print("Eligible to vote")
else:
  print("Not eligible to vote")
```

# **If elif else ladder**

```python
# creating grading system
# a >= 90 'A', 75 to 90 'B', 60 to 75 'C', 35 to 60 'D', 20 to 35 'E', 0 to 20 'F'

score = 701
if (score >=90) and (score <= 100):
  print("Grade A")
elif (score >=75) and (score < 90):
  print("Grade B")
elif (score >=60) and (score < 75):
  print("Grade C")
elif (score >=35) and (score < 60):
  print("Grade D")
elif (score >=20) and (score < 35):
  print("Grade E")
elif (score < 20) and (score >= 0):
  print("Grade F")
else:
  print("Invalid input")
```
