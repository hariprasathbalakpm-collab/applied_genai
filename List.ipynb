A List is the collection items stored in order, mutable and allows duplicate. Data stored in []

**Creating a list**

```python
fruits = ['apple','banana','cherry']
numbers = [1, 2, 5, 4, 3]
mixed = ['hello', 42, 3.14, True]

print(fruits)
print(numbers)
print(mixed)
```
