# **Collection types**

```python
# list
lst_1 =[1,"hari",1.2,True]
print(lst_1)
```

```python
# Tuples

tup_1 =(1,"hari", 2.1, False)
print(tup_1)
```

```python
# To call particular element from a list

print(lst_1[2])
print(lst_1[1])
```

```python
# To check datatype of a particular value in list

type(lst_1[0])
```

```python
# To update a value in list

lst_1[0] = 2
print(lst_1)
```

```python
tup_1[2]
```

```python
tup_1[2] = 'sethu'
```

```python
lst_1[1]='hari'

lst_1
```
